```python
# Reanalysis of the GlaMBIE project

This document describes the reanalysis performed on the paper by Zemp et al. (2025) published in Nature. We will introduce the paper, show how we reproduced their results and for our analysis explain (i) why we do each step, (ii) how we do it and (iii) what we learn from it.

## 1. Introduction of Zemp et al. (2025)

The reanalysis is performed on the paper “Community estimate of global glacier mass changes from 2000 to 20023” published by Zemp at al. (2025) in Nature. This study presents the results from the Glacier Mass Balance Intercomparison Exercise (GlaMBIE) which is a project collecting measurements of glacier mass balance from many different techniques from the research community. The paper aims to combine datasets from four observation methods (glaciological, DEM differencing, altimetry, gravimetry). To do this, they homogenize the data in space, time and unit, separate the temporal variability from the long term trend and produce combined estimates per region, method and globally.  
They find that global glacier mass decreased about \-6,542 \+/- 387 Gt from 2000 to 2023 (about \-273 \+/- 16 Gt/yr) which corresponds to a mass loss of about 5% globally or up to 39% regionally. They also report an increase in loss rate by 36 \+/- 10% from the first half (2000-2011) to the second half (2012-2023) of the examined period.

### Short overview of methods

 GlaMBIE combines 233 regional estimates of glacier mass changes from 35 research teams, covering 19 predefined glacier regions (based on the Randolf Glacier Inventory, RGI).

1. Homogenization: Datasets are converted to specific mass changes (m w.e.) using the glacier area which is derived from inventory data. Elevation-change methods assume an ice density for conversion (850 kg/m³ + conversion uncertainty). They are aligned to an annual grid by aggregating monthly or seasonal data to annual sums. Multi-annual data are corrected to hydrological year boundaries assuming linear change.

2. Separation of annual variability and trend: Each time series mean is subtracted. This isolates the annual anomaly from the long-term trend, avoiding artefacts from differing observation periods. The annual anomalies from multiple datasets within each method are averaged. The averaged anomaly is then re-trended by adding back the long-term trend (mean) of each individual input dataset, resulting in one time series per method. This produces one combined estimate: (i) glaciological + DEM differencing, (ii) altimetry, and (iii) gravimetry.

3. The previous step then is repeated for each method to obtain a single combined result per region. Lastly, those regional results are combined into a global estimate.

A benefit of this approach is that glaciological observations, which have a high interannual resolution, can be combined with DEM differencing, which provides precise estimates of overall mass changes.

Uncertainties are: (i) propagated observational uncertainties from input datasets, (ii) variability (spread) among input datasets within a method, and (iii) homogenization corrections considered (e.g., density assumption, temporal alignment).

```
